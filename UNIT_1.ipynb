```python
import numpy as np

A = np.array([2, 3, 4])
B = np.array([1, 2, 3])

print("Addition:", A + B)
print("Subtraction:", A - B)
print("Scalar Multiplication:", 2 * A)
```

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Multiplication:\n", np.dot(A, B))
print("Transpose:\n", A.T)
```

```python
import numpy as np

A = np.array([[2, 1], [5, 3]])

print("Determinant:", np.linalg.det(A))
print("Inverse:\n", np.linalg.inv(A))
print("Rank:", np.linalg.matrix_rank(A))
```

```python
import numpy as np

A = np.array([[0.5, 0.5],
              [0.2, 0.8]])

values, vectors = np.linalg.eig(A)

print("Eigenvalues:", values)
print("Eigenvectors:\n", vectors)
```

```python
import numpy as np

A = np.array([2, 3, 4])
B = np.array([3, 2, 4])

dot = np.dot(A, B)
cos = dot / (np.linalg.norm(A) * np.linalg.norm(B))

print("Dot Product:", dot)
print("Cosine Similarity:", cos)
```

```python
import numpy as np

A = np.array([2, 3])
B = np.array([5, 7])

print("Euclidean Distance:", np.linalg.norm(A-B))
print("Manhattan Distance:", np.sum(np.abs(A-B)))
```

```python
import numpy as np

marks = np.array([70, 80, 85, 90, 75])

print("Mean:", np.mean(marks))
print("Variance:", np.var(marks))
print("Standard Deviation:", np.std(marks))
```

```python
P_A = 0.6
P_B = 0.4

P_D_A = 0.02
P_D_B = 0.05

P_D = P_A*P_D_A + P_B*P_D_B

result = (P_A * P_D_A) / P_D

print("Probability:", result)
```

```python
prevalence = 0.01
sensitivity = 0.95
specificity = 0.90

p = (sensitivity * prevalence) / (
    sensitivity * prevalence +
    (1-specificity) * (1-prevalence)
)

print("Probability of Disease:", p)
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(30, 5, 1000)

print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))

plt.hist(data, bins=20, density=True)
plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Probability")
plt.show()
```

```python
import math

low = 6
high = 4
total = low + high

entropy = -(low/total)*math.log2(low/total) \
          -(high/total)*math.log2(high/total)

print("Entropy:", entropy)
```

```python
import numpy as np

actual = np.array([1, 0, 1, 1])
predicted = np.array([0.9, 0.2, 0.8, 0.7])

loss = -np.mean(
    actual*np.log(predicted) +
    (1-actual)*np.log(1-predicted)
)

print("Cross Entropy Loss:", loss)
```

```python
import numpy as np

P = np.array([0.4, 0.6])
Q = np.array([0.5, 0.5])

kl = np.sum(P * np.log(P/Q))

print("KL Divergence:", kl)
```

```python
x = 10
learning_rate = 0.1

for i in range(50):
    gradient = 2*x
    x = x - learning_rate*gradient

print("Minimum Value:", x)
print("Minimum Cost:", x**2)
```

```python
import numpy as np

data = np.array([2, 4, 6, 8])
lr = 0.01

for method in ["Batch", "Stochastic", "Mini-Batch"]:
    w = 0.0

    for epoch in range(20):
        if method == "Batch":
            batches = [data]
        elif method == "Stochastic":
            batches = [[x] for x in data]
        else:
            batches = [data[:2], data[2:]]

        for batch in batches:
            batch = np.array(batch)
            gradient = np.mean(w - batch)
            w = w - lr * gradient

    print(method, "Weight:", round(w, 3))
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

X = np.array([[1000, 2], [1500, 3],
              [2000, 3], [2500, 4],
              [3000, 4]])

y = np.array([20, 30, 40, 50, 60])

model = LinearRegression()
model.fit(X, y)

pred = model.predict(X)

print("Predicted Prices:", pred)
print("Mean Squared Error:", mean_squared_error(y, pred))
```
